1 Задача - Computing GC content

Входные данные:
Не более 10 строк ДНК в формате FASTA (длина каждой не более 1000 нуклеотидов).

Выходные данные:
Идентификатор строки с наибольшим GC-содержанием, за которым следует само значение GC-содержания в процентах. Rosalind допускает абсолютную погрешность 0.001.

```python
def count_CG(seq: str) -> float:
    '''
    :param seq: строка нуклеотидов
    :return: процент содержания C и G нуклеотидов
    '''
    assert isinstance(seq, str), "Входной параметр не является строкой"  # проверяем, что seq - строка
    CG_quantity = 0                       # счётчик: сколько нуклеотидов C или G встретилось
    for nucl in seq:                      # перебираем последовательность по одному символу
        if nucl in "CG":                  # проверяем: символ равен 'C' или 'G'?
            CG_quantity += 1              # если да - увеличиваем счётчик на 1
    return CG_quantity * 100 / len(seq)   # возвращаем процент: (CG / длина) * 100


with open("rosalind_gc.txt") as file:     # открываем файл
    content = file.readlines()            # читаем все строки файла в список

entries = {}                              # ключ словаря - это ID, значение - это сама последовательность
current_id = None                         # ID текущей проверяемой последовательности
for line in content:                      # проходим по каждой строке файла
    if line.startswith('>'):              # если строка начинается с '>' - это заголовок (ID)
        current_id = line.strip()[1:]     # отрезаем '>' - получаем чистый ID
        entries[current_id] = ""          # заводим для этого ID пустую последовательность
    else:                                 # иначе - это строка с частью последовательности
        entries[current_id] += line.strip()  # приклеиваем кусок к текущей последовательности

best_id = None                            # ID последовательности с максимальным GC
best_gc = -1.0                            # само значение GC; -1, чтобы любое реальное значение его превысило
for seq_id, seq in entries.items():       # перебираем пары (ID, последовательность) из словаря
    gc = count_CG(seq)                    # считаем GC-content для текущей последовательности
    if gc > best_gc:                      # если текущий GC больше лучшего -
        best_gc = gc                      # обновляем лучшее значение
        best_id = seq_id                  # и запоминаем ID этой последовательности

print(best_id)                            # печатаем ID последовательности с максимальным GC
print(f"{best_gc:.6f}")                   # печатаем сам GC с точностью 6 знаков после запятой
```

2 Задача - Counting Point Mutations

Входные данные:
Две строки ДНК s и t одинаковой длины (не более 1000 нуклеотидов).

Выходные данные:
Расстояние Хэмминга dH(s, t) — целое число, равное количеству позиций, в которых соответствующие символы строк s и t различаются.


```python
with open("rosalind_hamm.txt") as file:      # открываем файл
    s = file.readline().strip()              # читаем первую строку 
    t = file.readline().strip()              # читаем вторую строку

    distance = 0                             # счётчик различий 
    for i in range(len(s)):                  # идём по индексам от 0 до len(s)-1
        if s[i] != t[i]:                     # если символы на позиции i различаются -
            distance += 1                    # увеличиваем счётчик на 1

print(distance)                              # печатаем итоговое расстояние
```

3 Задача - Mendel's First Law

Входные данные:
Три положительных целых числа k, m и n, представляющих популяцию из k+m+n организмов: k особей — гомозиготные доминантные по определённому фактору, m — гетерозиготные, n — гомозиготные рецессивные.

Выходные данные:
Вероятность того, что два случайно выбранных для скрещивания организма произведут потомство, обладающее доминантным аллелем (и, следовательно, проявляющее доминантный фенотип). Предполагается, что любые два организма могут скрещиваться.

```python
with open("rosalind_iprb.txt") as file:              # открываем файл
    k, m, n = map(int, file.readline().split())      # читаем три числа из первой строки

# k - доминантные гомозиготы (AA)
# m - гетерозиготы (Aa)
# n - рецессивные гомозиготы (aa)

total = k + m + n                                    # общее число организмов в популяции

# Общее число пар родителей
total_cases = total * (total - 1)

#кол-во возможных пар родителей, где все потомки будут аа
homorecessive = n * (n - 1)

#кол-во возможных пар родителей - дающие рецессивного потомка
#Aa×Aa 
#Aa×aa 
heterorecessive = m * (m - 1) / 4 + m * n

p_rec = (homorecessive + heterorecessive) / total_cases   # вероятность рецессивного потомка
p_dom = 1 - p_rec                                          # вероятность доминантного фенотипа

print(p_dom)                                               # выводим результат
```

4 Задача - Majority Element

Входные данные:
Положительное целое число k ≤ 20, положительное целое число n ≤ 10⁴, и k массивов размера n, содержащих положительные целые числа, не превышающие 10⁵.

Выходные данные:
Для каждого массива вывести элемент, встречающийся строго более n/2 раз, если такой элемент существует, и "-1" в противном случае.

```python
def find_majority(nums: list, lo: int, hi: int):
    
    # nums: список целых чисел
    # lo:   левая граница (включительно)
    # hi:   правая граница (включительно)
    
    if lo == hi:                                        # базовый случай: один элемент -
        return nums[lo]                                 # он и есть «большинство» этой части

    mid = (lo + hi) // 2                                # середина отрезка

    left_cand = find_majority(nums, lo, mid)            # элемент из левой половины
    right_cand = find_majority(nums, mid + 1, hi)       # элемент из правой половины

    if left_cand == right_cand:                         # если элементы совпали - это ответ для целого
        return left_cand

    span = hi - lo + 1                                  # длина рассматриваемого отрезка

    # Считаем вхождения обоих элементов на отрезке за один проход
    left_hits = 0
    right_hits = 0
    for i in range(lo, hi + 1):
        if nums[i] == left_cand:
            left_hits += 1
        elif nums[i] == right_cand:
            right_hits += 1

    if left_hits * 2 > span:                            # строго больше половины
        return left_cand
    if right_hits * 2 > span:                           # строго больше половины
        return right_cand
    return -1                                           # большинства нет


answers = []                                            # сюда соберём ответы по каждому массиву
with open("rosalind_maj.txt") as file:                  # открываем файл
    num_arrays, size = map(int, file.readline().split())  # первая строка: сколько массивов и их длина
    for _ in range(num_arrays):                         # обрабатываем каждый массив
        arr = list(map(int, file.readline().split()))   # читаем строку в список целых
        answers.append(find_majority(arr, 0, size - 1)) # ищем большинство и добавляем в ответ

print(*answers)                                         
```

5 Задача - Merge Two Sorted Arrays 

Входные данные:
Положительное целое число n ≤ 10⁵ и отсортированный массив A[1..n] целых чисел от −10⁵ до 10⁵; положительное целое число m ≤ 10⁵ и отсортированный массив B[1..m] целых чисел от −10⁵ до 10⁵.

Выходные данные:
Отсортированный массив C[1..n+m], содержащий все элементы массивов A и B.

```python
def merge(arr1: list, arr2: list) -> list:
    
    #arr1: первый отсортированный список
    #arr2: второй отсортированный список
    
    assert isinstance(arr1, list) and isinstance(arr2, list), "Входные параметры не являются списками"
    res = []                                # пустой список, там будет результирующий список
    i, j = 0, 0                             # указатели по arr1 и arr2 
    n1, n2 = len(arr1), len(arr2)           # длины списков 
    while i < n1 and j < n2:                # цикл, пока существует оба списка (ни один не закончился)
        if arr1[i] <= arr2[j]:              # если первый элемент первого списка меньше или равен первому второго
            res.append(arr1[i])             # то добавляем тот, что меньше
            i += 1                          # и сдвигаем указатель первого списка
        else:
            res.append(arr2[j])             # иначе добавляем элемент второго списка
            j += 1                          # и сдвигаем указатель второго списка
    res.extend(arr1[i:])                    # если в первом списке ещё остались элементы, просто добавляю их в конец  списка
    res.extend(arr2[j:])                    # то же самое для второго списка, если в нём остались элемент
    return res


with open("rosalind_mer.txt") as file:              # открываем файл
    n = int(file.readline().strip())                # размер первой последовательности
    data_1 = list(map(int, file.readline().split()))  # первая последовательность чисел и преобразование в список

    m = int(file.readline().strip())                # считываю размер второй последовательности
    data_2 = list(map(int, file.readline().split()))  # вторая последовательность чисел и преобразование в список

print(*merge(data_1, data_2))                      
```

6 Задача - Translating RNA into Protein

Входные данные:
РНК-строка s, соответствующая цепи мРНК (длиной не более 10 kbp).

Выходные данные:
Белковая строка, закодированная строкой s.

```python
codon_table = {
    "UUU": "F", "CUU": "L", "AUU": "I", "GUU": "V",
    "UUC": "F", "CUC": "L", "AUC": "I", "GUC": "V",
    "UUA": "L", "CUA": "L", "AUA": "I", "GUA": "V",
    "UUG": "L", "CUG": "L", "AUG": "M", "GUG": "V",

    "UCU": "S", "CCU": "P", "ACU": "T", "GCU": "A",
    "UCC": "S", "CCC": "P", "ACC": "T", "GCC": "A",
    "UCA": "S", "CCA": "P", "ACA": "T", "GCA": "A",
    "UCG": "S", "CCG": "P", "ACG": "T", "GCG": "A",

    "UAU": "Y", "CAU": "H", "AAU": "N", "GAU": "D",
    "UAC": "Y", "CAC": "H", "AAC": "N", "GAC": "D",
    "UAA": "Stop", "CAA": "Q", "AAA": "K", "GAA": "E",
    "UAG": "Stop", "CAG": "Q", "AAG": "K", "GAG": "E",

    "UGU": "C", "CGU": "R", "AGU": "S", "GGU": "G",
    "UGC": "C", "CGC": "R", "AGC": "S", "GGC": "G",
    "UGA": "Stop", "CGA": "R", "AGA": "R", "GGA": "G",
    "UGG": "W", "CGG": "R", "AGG": "R", "GGG": "G"
}

with open("rosalind_prot.txt") as file:
    rna = file.read().strip()              # читаем всю РНК 

protein = []                               # список аминокислот белка
for i in range(0, len(rna), 3):            # идём по строке шагом 3 - по кодонам
    codon = rna[i:i + 3]                   # текущий триплет
    amino = codon_table[codon]             # аминокислота по таблице
    if amino == 'Stop':                    # дошли до стоп-кодона 
        break                              # завершаем трансляцию
    protein.append(amino)                  # иначе добавляем аминокислоту

print(''.join(protein))                    # склеиваем
```

7 Задача - Double-Degree Array

Входные данные:
Простой неориентированный граф с n ≤ 10³ вершинами в формате списка рёбер. Первая строка содержит два числа: n (количество вершин) и m (количество рёбер). Далее m строк, каждая содержит два числа — номера вершин, соединённых ребром.

Выходные данные:
Массив D[1..n], где D[i] — сумма степеней всех соседей вершины i.
